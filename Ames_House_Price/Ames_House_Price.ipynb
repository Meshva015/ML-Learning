## Data Preprocessing

### Separate Features and Target

SalePrice is the target variable, while the remaining columns are used as input features.

```python
X = df.drop(columns=["SalePrice", "LogSalePrice"])
y = np.log1p(df["SalePrice"])

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

### Train-Test Split

The dataset is split into training and testing sets before preprocessing and model fitting.
An 80/20 split is used with a fixed random state for reproducibility.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

### Missing Value 


```python
absent_columns = [ "PoolQC", "Alley", "Fence", "FireplaceQu", "GarageType", "GarageFinish", "GarageQual", "GarageCond", "BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1", "BsmtFinType2", "MiscFeature"]

print(absent_columns)
```

```python
genuine_missing_columns = ["LotFrontage", "MasVnrArea", "Electrical"]

print("Genuinely missing columns:")
print(genuine_missing_columns)
```

```python
for col in absent_columns:
    X_train[col] = X_train[col].fillna("None")
    X_test[col] = X_test[col].fillna("None")

print("Feature-absence missing values handled.")
```

```python
genuine_numeric = ["LotFrontage", "MasVnrArea"]

for col in genuine_numeric:
    median_value = X_train[col].median()
    
    X_train[col] = X_train[col].fillna(median_value)
    X_test[col] = X_test[col].fillna(median_value)

electrical_mode = X_train["Electrical"].mode()[0]

X_train["Electrical"] = X_train["Electrical"].fillna(electrical_mode)
X_test["Electrical"] = X_test["Electrical"].fillna(electrical_mode)

print("Genuine missing values handled.")
```

```python
print("Missing values in X_train:")
print(X_train.isnull().sum()[X_train.isnull().sum() > 0])

print("\nMissing values in X_test:")
print(X_test.isnull().sum()[X_test.isnull().sum() > 0])
```

Ordinal categorical features have a natural order.

```python
for col in ordinal_columns:
    X_train[col] = X_train[col].map(ordinal_mapping)
    X_test[col] = X_test[col].map(ordinal_mapping)

print("Ordinal encoding completed.")
```

### Skewed Numeric Features

Highly skewed numeric features are identified using skewness.
A log1p transformation can reduce right skewness and make the feature distribution more suitable for regression models.

```python
numeric_columns = X_train.select_dtypes(include=np.number).columns.tolist()

numeric_columns = [
    col for col in numeric_columns
    if col not in ordinal_columns
]

print("Number of numeric columns:", len(numeric_columns))
```

### Target 

The SalePrice target is transformed using log1p to reduce its right skewness.
Predictions can later be converted back to the original price scale using expm1.

```python
print("Original SalePrice example:")
print(df["SalePrice"].head())

print("\nLog-transformed target example:")
print(y.head())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder

numeric_columns = X_train.select_dtypes(include=np.number).columns.tolist()

numeric_columns = [
    col for col in numeric_columns
    if col not in ordinal_columns
]

nominal_columns = [
    col for col in X_train.columns
    if col not in numeric_columns
    and col not in ordinal_columns
]

print("Numeric columns:", len(numeric_columns))
print("Ordinal columns:", len(ordinal_columns))
print("Nominal columns:", len(nominal_columns))
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            SimpleImputer(strategy="median"),
            numeric_columns
        ),
        (
            "ordinal",
            SimpleImputer(strategy="most_frequent"),
            ordinal_columns
        ),
        (
            "nominal",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("onehot", OneHotEncoder(
                    handle_unknown="ignore",
                    sparse_output=False
                ))
            ]),
            nominal_columns
        )
    ],
    remainder="drop"
)

print("Final preprocessor created successfully.")
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("X_train processed shape:", X_train_processed.shape)
print("X_test processed shape:", X_test_processed.shape)
```

## Feature Engineering

Feature engineering creates new features from existing columns to capture useful relationships in the data.

### TotalSF

TotalSF combines basement, first-floor, and second-floor living areas to represent the total house area.

```python
X_train["TotalSF"] = (
    X_train["TotalBsmtSF"]
    + X_train["1stFlrSF"]
    + X_train["2ndFlrSF"]
)

X_test["TotalSF"] = (
    X_test["TotalBsmtSF"]
    + X_test["1stFlrSF"]
    + X_test["2ndFlrSF"]
)
```

### TotalBath

TotalBath combines full and half bathrooms, including basement bathrooms, to represent the total bathroom capacity of the house.

```python
X_train["TotalBath"] = (X_train["FullBath"] + 0.5 * X_train["HalfBath"] +
                         X_train["BsmtFullBath"] + 0.5 * X_train["BsmtHalfBath"])
X_test["TotalBath"] = (X_test["FullBath"] + 0.5 * X_test["HalfBath"] +
                        X_test["BsmtFullBath"] + 0.5 * X_test["BsmtHalfBath"])
```

### HouseAge

HouseAge represents the age of the house at the time it was sold.

```python
X_train["HouseAge"] = X_train["YrSold"].astype(int) - X_train["YearBuilt"]
X_test["HouseAge"] = X_test["YrSold"].astype(int) - X_test["YearBuilt"]
```

```python
X_train[["YrSold", "YearBuilt", "HouseAge"]].head()
```

### YearsSinceRemodel

YearsSinceRemodel represents the number of years since the house was last remodeled when it was sold.

```python
X_train["YearsSinceRemodel"] = X_train["YrSold"].astype(int) - X_train["YearRemodAdd"]
X_test["YearsSinceRemodel"] = X_test["YrSold"].astype(int) - X_test["YearRemodAdd"]
```

### TotalPorchSF

TotalPorchSF combines different porch and deck areas into one feature representing the total outdoor living area.

```python
X_train["TotalPorchSF"] = (X_train["OpenPorchSF"] + X_train["EnclosedPorch"] +
                            X_train["3SsnPorch"] + X_train["ScreenPorch"])
X_test["TotalPorchSF"] = (X_test["OpenPorchSF"] + X_test["EnclosedPorch"] +
                           X_test["3SsnPorch"] + X_test["ScreenPorch"])
```

### Binary Features

Binary features indicate whether a house has specific facilities such as a pool, garage, basement, or fireplace.

```python
for train_test_df in [X_train, X_test]:
    train_test_df["HasPool"] = (train_test_df["PoolArea"] > 0).astype(int)
    train_test_df["HasGarage"] = (train_test_df["GarageArea"] > 0).astype(int)
    train_test_df["HasBasement"] = (train_test_df["TotalBsmtSF"] > 0).astype(int)
    train_test_df["HasFireplace"] = (train_test_df["Fireplaces"] > 0).astype(int)
    train_test_df["IsRemodeled"] = (train_test_df["YearBuilt"] != train_test_df["YearRemodAdd"]).astype(int)
```

### interaction feature (OverallQual × GrLivArea)

```python
X_train["OverallQual_GrLivArea"] = (
    X_train["OverallQual"] * X_train["GrLivArea"]
)

X_test["OverallQual_GrLivArea"] = (
    X_test["OverallQual"] * X_test["GrLivArea"]
)
```

### Learning Curves

Learning curves show how model performance changes with different training-set sizes and help identify underfitting or overfitting.

```python
from sklearn.model_selection import learning_curve

models_for_learning = {
    "Lasso Regression": lasso_model,
    "KNN Regression": knn_model
}

for name, model in models_for_learning.items():

    train_sizes, train_scores, validation_scores = learning_curve(
        model,
        X_train_processed,
        y_train,
        cv=5,
        scoring="neg_root_mean_squared_error",
        train_sizes=np.linspace(0.1, 1.0, 5),
        n_jobs=-1
    )

    train_rmse = -train_scores.mean(axis=1)
    validation_rmse = -validation_scores.mean(axis=1)

    plt.figure(figsize=(8, 5))
    plt.plot(train_sizes, train_rmse, marker="o", label="Training RMSE")
    plt.plot(train_sizes, validation_rmse, marker="o", label="Validation RMSE")

    plt.title(f"Learning Curve - {name}")
    plt.xlabel("Training Samples")
    plt.ylabel("RMSE (Log SalePrice)")
    plt.legend()
    plt.grid(True)
    plt.show()
```
