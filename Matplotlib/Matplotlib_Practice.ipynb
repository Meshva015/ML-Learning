```python
import matplotlib.pyplot as plt
```

```python
x = [10, 20, 30, 40, 50, 60]
y = [13, 45, 23, 34, 96, 76]

plt.title("Bar Graph")
plt.bar(x, y, color='dodgerblue', width=5)
plt.show()
```

```python
# Pie Chart Example

subjects = ["Python", "Java", "SQL", "ML"]
hours = [10, 6, 8, 12]

plt.pie(hours, labels=subjects, autopct="%1.1f%%")

plt.title("Study Hours by Subject")

plt.show()
```

```python
# Line Chart Example

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
sales = [20, 35, 30, 45, 50]

plt.plot(days, sales, marker="o")

plt.title("Weekly Sales")
plt.xlabel("Days")
plt.ylabel("Sales")

plt.show()
```

```python
# Scatter Plot Example

hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [45, 50, 55, 60, 65, 72, 78, 85]

plt.scatter(hours, marks)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
# Histogram Example

marks = [45, 50, 52, 55, 58, 60, 62, 65, 67, 70,
         72, 75, 78, 80, 82, 85, 88, 90, 92, 95]

plt.hist(marks, bins=5)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Box Plot Example

marks = [45, 50, 52, 55, 58, 60, 62, 65, 67, 70,
         72, 75, 78, 80, 82, 85, 88, 90, 92, 95]

plt.boxplot(marks)

plt.title("Distribution of Student Marks")
plt.ylabel("Marks")

plt.show()
```

```python
# Subplots Example

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
sales = [20, 35, 30, 45, 50]
expenses = [15, 20, 25, 30, 35]

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(days, sales, marker="o")
plt.title("Weekly Sales")
plt.xlabel("Days")
plt.ylabel("Sales")

plt.subplot(1, 2, 2)
plt.bar(days, expenses)
plt.title("Weekly Expenses")
plt.xlabel("Days")
plt.ylabel("Expenses")

plt.tight_layout()
plt.show()
```
