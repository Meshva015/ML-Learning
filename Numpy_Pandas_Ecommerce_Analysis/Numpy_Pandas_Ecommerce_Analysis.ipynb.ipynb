```python
import pandas as pd
import numpy as np
```

```python
customers = pd.read_csv("customers.csv")
products = pd.read_csv("products.csv")
orders = pd.read_csv("orders.csv")
```

```python
print("Customers Shape:", customers.shape)
print("Products Shape:", products.shape)
print("Orders Shape:", orders.shape)
```

```python
print("Customers Columns:")
print(customers.columns)

print("\nProducts Columns:")
print(products.columns)

print("\nOrders Columns:")
print(orders.columns)
```

```python
print("Customers Data Types:")
print(customers.dtypes)

print("\nProducts Data Types:")
print(products.dtypes)

print("\nOrders Data Types:")
print(orders.dtypes)
```

```python
print("Customers Missing Values:")
print(customers.isnull().sum())

print("\nProducts Missing Values:")
print(products.isnull().sum())

print("\nOrders Missing Values:")
print(orders.isnull().sum())
```

```python
print("Customers Duplicate Rows:", customers.duplicated().sum())
print("Products Duplicate Rows:", products.duplicated().sum())
print("Orders Duplicate Rows:", orders.duplicated().sum())
```

```python
customers = customers.drop_duplicates()
products = products.drop_duplicates()
orders = orders.drop_duplicates()
```

```python
print("Customers Duplicate Rows:", customers.duplicated().sum())
print("Products Duplicate Rows:", products.duplicated().sum())
print("Orders Duplicate Rows:", orders.duplicated().sum())
```

```python
print("Customers Missing Email:")
print(customers[customers["email"].isnull()])

print("\nProducts Missing Price:")
print(products[products["price"].isnull()])

print("\nOrders Missing Payment Method:")
print(orders[orders["payment_method"].isnull()])
```

```python
print(products)
```

```python
# Handle missing email
customers["email"] = customers["email"].fillna("Not Available")

# Handle missing price using median
products["price"] = products["price"].fillna(products["price"].median())

# Handle missing payment method
orders["payment_method"] = orders["payment_method"].fillna("Unknown")
```

```python
print("Customers Missing Values:")
print(customers.isnull().sum())

print("\nProducts Missing Values:")
print(products.isnull().sum())

print("\nOrders Missing Values:")
print(orders.isnull().sum())
```

```python
# Customers
customers["customer_name"] = customers["customer_name"].str.strip().str.title()
customers["email"] = customers["email"].str.strip()
customers["city"] = customers["city"].str.strip().str.title()
customers["gender"] = customers["gender"].str.strip().str.title()

# Products
products["product_name"] = products["product_name"].str.strip().str.title()
products["category"] = products["category"].str.strip().str.title()

# Orders
orders["payment_method"] = orders["payment_method"].str.strip().str.title()
orders["order_status"] = orders["order_status"].str.strip().str.title()
```

```python
# Save cleaned datasets

customers.to_csv("customers_cleaned.csv", index=False)
products.to_csv("products_cleaned.csv", index=False)
orders.to_csv("orders_cleaned.csv", index=False)

print("Cleaned datasets saved successfully!")
```

```python
# Merge orders with products to get product price

orders_with_price = orders.merge(
    products[["product_id", "price"]],
    on="product_id",
    how="left"
)

# Calculate total price for each order
orders_with_price["total_price"] = (
    orders_with_price["quantity"] * orders_with_price["price"]
)

# Calculate total spending for each customer
customer_spending = (
    orders_with_price.groupby("customer_id")["total_price"]
    .sum()
)

# Get top 5 customers
top_5_customers = customer_spending.sort_values(
    ascending=False
).head(5)

print("Top 5 Customers by Total Spending:")
print(top_5_customers)
```

```python
# Find the most sold product

product_sales = (
    orders.groupby("product_id")["quantity"]
    .sum()
    .sort_values(ascending=False)
)

most_sold_product = product_sales.head(1)

print("Most Sold Product:")
print(most_sold_product)
```

```python
# Merge orders with products to get category and price

orders_with_product = orders.merge(
    products[["product_id", "category", "price"]],
    on="product_id",
    how="left"
)

# Calculate revenue for each order
orders_with_product["revenue"] = (
    orders_with_product["quantity"] * orders_with_product["price"]
)

# Calculate total revenue by category
category_revenue = (
    orders_with_product.groupby("category")["revenue"]
    .sum()
    .sort_values(ascending=False)
)

print("Revenue by Category:")
print(category_revenue)

print("\nHighest Revenue Category:")
print(category_revenue.head(1))
```

```python
# Calculate average order quantity

average_order_quantity = orders["quantity"].mean()

print("Average Order Quantity:", average_order_quantity)
```

```python
# Find the city with the most customers

city_customer_count = (
    customers.groupby("city")["customer_id"]
    .count()
    .sort_values(ascending=False)
)

print("Customers by City:")
print(city_customer_count)

print("\nCity with Most Customers:")
print(city_customer_count.head(1))
```

```python
# Create Total Price column

orders_with_product["total_price"] = (
    orders_with_product["quantity"] * orders_with_product["price"]
)

print(orders_with_product[[
    "order_id",
    "product_id",
    "quantity",
    "price",
    "total_price"
]].head())
```

```python
# Create Discounted Price column

orders_with_product["discounted_price"] = (
    orders_with_product["total_price"]
    * (1 - orders_with_product["discount"] / 100)
)

print(orders_with_product[[
    "order_id",
    "quantity",
    "price",
    "discount",
    "total_price",
    "discounted_price"
]].head())
```

```python
# Calculate product-wise sales

product_wise_sales = (
    orders.groupby("product_id")["quantity"]
    .sum()
    .sort_values(ascending=False)
)

print("Product-wise Sales:")
print(product_wise_sales)
```

```python
# Calculate customer-wise spending

customer_wise_spending = (
    orders_with_product.groupby("customer_id")["total_price"]
    .sum()
    .sort_values(ascending=False)
)

print("Customer-wise Spending:")
print(customer_wise_spending)
```

```python
# Convert order_date to datetime

orders_with_product["order_date"] = pd.to_datetime(
    orders_with_product["order_date"]
)

# Extract month

orders_with_product["month"] = (
    orders_with_product["order_date"].dt.to_period("M")
)

# Calculate monthly sales

monthly_sales = (
    orders_with_product.groupby("month")["total_price"]
    .sum()
)

print("Monthly Sales:")
print(monthly_sales)
```

```python
# Filter orders with discount greater than 10%

high_discount_orders = orders[orders["discount"] > 10]

print("Orders with Discount Greater Than 10%:")
print(high_discount_orders)
```

```python
# Filter only delivered orders

delivered_orders = orders[orders["order_status"] == "Delivered"]

print("Delivered Orders:")
print(delivered_orders)
```

```python
# Filter customers from Ahmedabad

ahmedabad_customers = customers[
    customers["city"].str.strip().str.lower() == "ahmedabad"
]

print("Ahmedabad Customers:")
print(ahmedabad_customers)
```

```python
# Sort customers by highest total spending

sorted_customers = customer_wise_spending.sort_values(ascending=False)

print("Customers Sorted by Highest Spending:")
print(sorted_customers)
```

```python
# Sort products by highest price

most_expensive_products = products.sort_values(
    by="price",
    ascending=False
)

print("Most Expensive Products:")
print(most_expensive_products[[
    "product_id",
    "product_name",
    "category",
    "price"
]])
```

```python
# Convert customer names to uppercase

customers["customer_name_upper"] = customers["customer_name"].str.upper()

print("Customer Names in Uppercase:")
print(customers[["customer_name", "customer_name_upper"]])
```

```python
# Convert order_date to datetime format

orders["order_date"] = pd.to_datetime(orders["order_date"])

# Extract month, year and weekday
orders["month"] = orders["order_date"].dt.month
orders["year"] = orders["order_date"].dt.year
orders["weekday"] = orders["order_date"].dt.day_name()

print("Order Date with Month, Year and Weekday:")
print(orders[[
    "order_date",
    "month",
    "year",
    "weekday"
]])
```
