# Types of Machine Learning

Machine Learning is mainly divided into three types:

## 1. Supervised Learning

In supervised learning, the model learns from **labeled data**.

- Input data (X) is given.
- Correct output/target (y) is also given.
- The model learns the relationship between input and output.
- It is mainly used for prediction.

### Types:
- Classification → Predicts categories/classes
- Regression → Predicts continuous numerical values

### Examples:
- Spam or Not Spam → Classification
- House Price Prediction → Regression
- Disease Prediction → Classification

### Algorithms:
- Linear Regression
- Logistic Regression
- KNN
- SVM
- Decision Tree
- Random Forest
- Naive Bayes

---

## 2. Unsupervised Learning

In unsupervised learning, the model learns from **unlabeled data**.

- Only input data (X) is given.
- There is no target/output column.
- The model tries to find hidden patterns or groups in the data.

### Main Types:
- Clustering
- Dimensionality Reduction

### Examples:
- Customer Segmentation
- Grouping similar products
- Finding similar users

### Algorithms:
- K-Means Clustering
- Hierarchical Clustering
- DBSCAN
- PCA

---

## 3. Reinforcement Learning

In reinforcement learning, an **agent learns by interacting with an environment**.

The agent:
- Takes an action
- Receives a reward or penalty
- Learns which actions give better results

### Examples:
- Game-playing AI
- Robot navigation
- Self-driving systems
- Recommendation systems

### Basic Terms:
- Agent → Learner/decision maker
- Environment → World in which agent operates
- Action → Decision taken by agent
- Reward → Feedback received after action

---

## Quick Comparison

| Type | Data | Target | Main Purpose |
|------|------|--------|--------------|
| Supervised | Labeled | Available | Prediction |
| Unsupervised | Unlabeled | Not available | Find patterns |
| Reinforcement | Interaction/feedback | Reward | Learn best actions |

# Types of Data Required in Different Types of ML

Different Machine Learning techniques require different types of data.

## 1. Supervised Learning

### Required Data:
Labeled data

The dataset contains:
- Features/Input (X)
- Target/Output (y)

### Example:

| Hours_Studied | Attendance | Result |
|---|---:|---|
| 2 | 60 | Fail |
| 5 | 80 | Pass |
| 8 | 90 | Pass |

Here:
- Hours_Studied and Attendance → Features
- Result → Target/Label

### Used For:
- Classification
- Regression

---

## 2. Unsupervised Learning

### Required Data:
Unlabeled data

There is no target column.

### Example:

| Age | Income | Spending_Score |
|---:|---:|---:|
| 22 | 25000 | 80 |
| 45 | 70000 | 30 |
| 25 | 30000 | 75 |

The model tries to find groups/patterns by itself.

### Used For:
- Clustering
- Dimensionality Reduction

---

## 3. Reinforcement Learning

Reinforcement Learning does not require a traditional labeled dataset.

Instead, the model learns through:
- Environment
- Actions
- Rewards
- Penalties
- Feedback

### Example:

A robot learns to reach a destination.

- Correct movement → Positive reward
- Wrong movement → Negative reward

The agent learns the best actions through repeated interaction.

# Difference Between Machine Learning and Deep Learning

## Machine Learning (ML)

Machine Learning is a subset of Artificial Intelligence where algorithms learn patterns from data and make predictions or decisions.

Examples:
- Linear Regression
- Logistic Regression
- KNN
- SVM
- Decision Tree
- Random Forest
- Naive Bayes

---

## Deep Learning (DL)

Deep Learning is a subset of Machine Learning that uses **Artificial Neural Networks with multiple layers** to learn complex patterns from large amounts of data.

Examples:
- CNN
- RNN
- LSTM
- Transformer
- Deep Neural Networks

---

## ML vs DL

| Feature | Machine Learning | Deep Learning |
|---|---|---|
| Meaning | Algorithms learn patterns from data | Neural networks learn complex patterns |
| Data Requirement | Can work with smaller datasets | Usually requires large amounts of data |
| Feature Engineering | Often required | Often learns features automatically |
| Computing Power | Usually lower | Usually higher |
| Hardware | CPU is often sufficient | GPU is commonly useful |
| Training Time | Generally faster | Generally slower |
| Model Complexity | Lower | Higher |
| Examples | KNN, SVM, Random Forest | CNN, RNN, LSTM, Transformer |

---
