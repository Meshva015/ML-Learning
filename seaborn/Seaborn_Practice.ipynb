# Seaborn Practice

Seaborn is a Python data visualization library based on Matplotlib.

It provides a high-level interface for creating attractive and informative statistical graphics.

In this notebook, we will practice different Seaborn plots such as:

- Line Plot
- Scatter Plot
- Bar Plot
- Box Plot
- Histogram
- Heatmap
- Pair Plot

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 9],
    "Marks": [45, 50, 55, 65, 70, 78, 85, 92],
    "Gender": ["F", "M", "F", "M", "F", "M", "F", "M"]
}

df = pd.DataFrame(data)

df
```

## 1. Line Plot

A line plot is used to show trends or changes in data.

Here, we will visualize the relationship between Study Hours and Marks.

```python
sns.lineplot(x="Study_Hours", y="Marks", data=df)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

## 2. Scatter Plot

A scatter plot is used to show the relationship between two numerical variables.

Here, we will visualize the relationship between Study Hours and Marks.

```python
sns.scatterplot(x="Study_Hours", y="Marks", data=df)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

## 3. Bar Plot

A bar plot is used to compare values across different categories.

Here, we will compare the average marks of different genders.

```python
sns.barplot(x="Gender", y="Marks", data=df)

plt.title("Average Marks by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Marks")

plt.show()
```

## 4. Box Plot

A box plot is used to visualize the distribution of numerical data.

It helps to understand the median, quartiles, spread, and possible outliers.

```python
sns.boxplot(x="Gender", y="Marks", data=df)

plt.title("Marks Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Marks")

plt.show()
```

## 5. Histogram

A histogram is used to understand the distribution of a numerical variable.

Here, we will visualize the distribution of student marks.

```python
sns.histplot(df["Marks"], kde=True)

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

## 6. Heatmap

A heatmap represents numerical values using different levels of shading.

It is commonly used to visualize correlation between numerical variables.

```python
corr = df[["Study_Hours", "Marks"]].corr()

corr
```

```python
sns.heatmap(corr, annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

## 7. Pair Plot

A pair plot is used to visualize pairwise relationships between multiple numerical variables.

It creates a grid of plots for the selected variables.

```python
sns.pairplot(
    df,
    vars=["Study_Hours", "Marks"],
    hue="Gender"
)

plt.show()
```

## Seaborn Plot Summary

| Plot | Main Purpose |
|---|---|
| Line Plot | Show trends and changes |
| Scatter Plot | Show relationship between numerical variables |
| Bar Plot | Compare categories |
| Box Plot | Analyze distribution and outliers |
| Histogram | Analyze numerical data distribution |
| Heatmap | Visualize correlation or matrix data |
| Pair Plot | Explore relationships between multiple variables |
